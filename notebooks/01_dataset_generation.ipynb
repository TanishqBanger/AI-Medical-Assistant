```python
# -------------------------------------------------------
# Enhance department configuration
# -------------------------------------------------------

default_gender = ["Male", "Female", "Other"]

default_severity = {
    "Mild": 0.45,
    "Moderate": 0.40,
    "Severe": 0.15
}

for dept, config in departments.items():

    # Allowed genders
    config["allowed_gender"] = (
        ["Female"] if dept == "Gynecology" else default_gender
    )

    # Chronic condition probabilities
    if dept == "Cardiology":
        config["chronic_conditions"] = {
            "Hypertension": 0.40,
            "Heart Disease": 0.30,
            "Diabetes": 0.20,
            "None": 0.10
        }

    elif dept == "Endocrinology":
        config["chronic_conditions"] = {
            "Diabetes": 0.70,
            "None": 0.30
        }

    elif dept == "Pulmonology":
        config["chronic_conditions"] = {
            "Asthma": 0.60,
            "None": 0.40
        }

    elif dept == "Gynecology":
        config["chronic_conditions"] = {
            "PCOS": 0.45,
            "None": 0.55
        }

    elif dept == "Orthopedics":
        config["chronic_conditions"] = {
            "Arthritis": 0.40,
            "None": 0.60
        }

    elif dept in ["General Medicine", "Neurology"]:
        config["chronic_conditions"] = {
            "None": 0.50,
            "Hypertension": 0.25,
            "Diabetes": 0.25
        }

    else:
        config["chronic_conditions"] = {
            "None": 0.70,
            "Diabetes": 0.30
        }

    # Severity distribution
    if dept == "Cardiology":
        config["severity_distribution"] = {
            "Mild": 0.10,
            "Moderate": 0.45,
            "Severe": 0.45
        }

    elif dept == "Pulmonology":
        config["severity_distribution"] = {
            "Mild": 0.20,
            "Moderate": 0.45,
            "Severe": 0.35
        }

    elif dept == "Dermatology":
        config["severity_distribution"] = {
            "Mild": 0.65,
            "Moderate": 0.30,
            "Severe": 0.05
        }

    elif dept == "Psychiatry":
        config["severity_distribution"] = {
            "Mild": 0.30,
            "Moderate": 0.50,
            "Severe": 0.20
        }

    else:
        config["severity_distribution"] = default_severity




# -------------------------------------------------------
# Overlapping symptoms across departments
# -------------------------------------------------------

overlapping_symptoms = {
    "headache": ["General Medicine", "Neurology"],
    "cough": ["General Medicine", "Pulmonology"],
    "fever": ["General Medicine", "Pediatrics"],
    "fatigue": ["General Medicine", "Endocrinology"],
    "dizziness": ["Cardiology", "Neurology"],
    "shortness of breath": ["Cardiology", "Pulmonology"],
    "vomiting": ["Gastroenterology", "Pediatrics"],
    "body ache": ["General Medicine", "Orthopedics"],
    "swelling": ["Dermatology", "Orthopedics"],
    "pain": ["Dentistry", "Orthopedics", "Cardiology"],
}


severity_profiles = {
    "Mild": {
        "pain": (1, 3),
        "temp_offset": "low",
        "duration": (1, 5)
    },
    "Moderate": {
        "pain": (4, 6),
        "temp_offset": "medium",
        "duration": (3, 10)
    },
    "Severe": {
        "pain": (7, 10),
        "temp_offset": "high",
        "duration": (7, 30)
    }
}
```

#Funtions

```python
def generate_age(department):
    """
    Generate age based on department.
    """
    min_age, max_age = departments[department]["age_range"]
    return random.randint(min_age, max_age)


def generate_gender(department):
    """
    Generate gender based on department.
    """
    allowed = departments[department]["allowed_gender"]
    return random.choice(allowed)


def generate_symptoms(department):

    # Select one common symptom pattern
    symptoms = random.choice(symptom_patterns[department]).copy()

    # 30% chance to add one extra symptom from the same department
    if random.random() < 0.30:

        available = list(
            set(departments[department]["symptoms"]) - set(symptoms)
        )

        if available:
            symptoms.append(random.choice(available))

    # 20% chance to add one overlapping symptom
    if random.random() < 0.20:

        overlap_candidates = []

        for symptom, depts in overlapping_symptoms.items():
            if department in depts:
                overlap_candidates.append(symptom)

        if overlap_candidates:

            symptom = random.choice(overlap_candidates)

            if symptom not in symptoms:
                symptoms.append(symptom)

    return ", ".join(symptoms)


def generate_temperature(department, severity):

    low, high = departments[department]["temperature_range"]

    if severity == "Mild":
        temp = random.uniform(low, min(low + 1.5, high))

    elif severity == "Moderate":
        temp = random.uniform(low + 1.0, min(low + 3.0, high))

    else:
        temp = random.uniform(max(high - 1.5, low), high)

    return round(temp, 1)


def generate_bp(department, severity):
    s_low, s_high = departments[department]["bp_systolic"]

    # Generate systolic
    systolic = random.randint(s_low, s_high)

    # Generate diastolic relative to systolic
    diastolic = systolic - random.randint(35, 55)

    # Keep within department limits
    d_low, d_high = departments[department]["bp_diastolic"]
    diastolic = max(d_low, min(diastolic, d_high))

    return systolic, diastolic


def generate_heart_rate(department, severity):

    low, high = departments[department]["heart_rate"]

    return random.randint(low, high)



def generate_oxygen(department, severity):

    low, high = departments[department]["oxygen"]

    return random.randint(low, high)


def generate_pain(department, severity):

    # Department pain range
    low, high = departments[department]["pain_level"]

    # Number of pain values in this department
    total_values = high - low + 1

    # Size of one severity block
    block = max(1, total_values // 3)

    if severity == "Mild":
        start = low
        end = min(low + block - 1, high)

    elif severity == "Moderate":
        start = min(low + block, high)
        end = min(low + (2 * block) - 1, high)

    else:   # Severe
        start = min(low + (2 * block), high)
        end = high

    return random.randint(start, end)


def generate_duration(severity):

    low, high = severity_profiles[severity]["duration"]

    return random.randint(low, high)


def generate_severity(department):

    severity = departments[department]["severity_distribution"]

    return random.choices(
        population=list(severity.keys()),
        weights=list(severity.values()),
        k=1
    )[0]



def generate_chronic_condition(department):

    conditions = departments[department]["chronic_conditions"]

    return random.choices(
        population=list(conditions.keys()),
        weights=list(conditions.values()),
        k=1
    )[0]


def generate_patient_id(index):

    return 100000 + index


def generate_department():

    return random.choice(
        list(departments.keys())
    )



# Testing the functions

department = generate_department()

severity = generate_severity(department)

print("Department:", department)
print("Age:", generate_age(department))
print("Gender:", generate_gender(department))
print("Symptoms:", generate_symptoms(department))
print("Severity:", severity)
print("Temperature:", generate_temperature(department, severity))
print("Blood Pressure:", generate_bp(department, severity))
print("Heart Rate:", generate_heart_rate(department, severity))
print("Oxygen:", generate_oxygen(department, severity))
print("Pain:", generate_pain(department, severity))
print("Chronic:", generate_chronic_condition(department))

```

# generating patient

```python
def generate_patient(patient_id):
    """
    Generate one synthetic patient record.
    """

    # Step 1: Select Department
    department = generate_department()

    # Step 2: Demographics
    age = generate_age(department)
    gender = generate_gender(department)

    # Step 3: Clinical Information
    severity = generate_severity(department)
    symptoms = generate_symptoms(department)
    temperature = generate_temperature(department, severity)
    duration = generate_duration(severity)

    # Step 4: Vital Signs
    systolic_bp, diastolic_bp = generate_bp(department, severity)
    heart_rate = generate_heart_rate(department, severity)
    oxygen = generate_oxygen(department, severity)
    pain = generate_pain(department, severity)

    # Step 5: Medical History
    chronic_condition = generate_chronic_condition(department)

    # Step 6: Return Patient Record
    patient = {
        "patient_id": patient_id,
        "age": age,
        "gender": gender,
        "symptoms": symptoms,
        "temperature": temperature,
        "duration_days": duration,
        "blood_pressure_systolic": systolic_bp,
        "blood_pressure_diastolic": diastolic_bp,
        "heart_rate": heart_rate,
        "oxygen_saturation": oxygen,
        "pain_level": pain,
        "symptom_severity": severity,
        "chronic_condition": chronic_condition,
        "department": department
    }

    return patient


patient = generate_patient(100001)

print(patient)
```
